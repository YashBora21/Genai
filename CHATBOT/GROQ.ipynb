```python
### messgae History 
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history  import BaseChatMessageHistory
from langchain_core.runnables.history  import RunnableWithMessageHistory
store={}
def session_history_per_user (session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]
with_message_history=RunnableWithMessageHistory(model,session_history_per_user)
```

```python
config={"configurable":{"session_id":"chat1"}}
```

```python
with_message_history.invoke(
    [HumanMessage(content="Hey I am yash And I an Data Scientist")],
    config=config #store history for chat 1 user
)
```

```python
with_message_history.invoke(
    [HumanMessage(content="whats my name")],
    config=config #store history for chat 1 user
)
```

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder

prompt=ChatPromptTemplate.from_messages(
    [
        ("system","Hey You are a Helpful Assitant Answer to all my question"),
        MessagesPlaceholder(variable_name="message"
        )
    ]
)
chain=prompt|model
chain.invoke({"message":[HumanMessage(content="Hey I am Yash ")]})
```

```python
with_message_history=RunnableWithMessageHistory(chain,session_history_per_user)
```

```python
config1={"configurable":{"session_id":"chat4"}}
with_message_history.invoke(
    HumanMessage(content="Hey I am Yash "),
    config=config1 #store history for chat 1 user
)
```

```python
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","Hey You are a Helpful Assitant Answer to all my question in this {language}"),
        MessagesPlaceholder(variable_name="message"
        )
    ]
)
chain=prompt|model
chain.invoke({"message":[HumanMessage(content="Hey I am Yash ")],"language":"egyptian"})
```

```python
def local_token_counter(messages):
#      Rough estimation: 1 token ≈ 4 characters or 0.75 words
  return sum(len(msg.content.split()) for msg in messages)
```

```python
trimmer=trim_messages(
    max_tokens=200,
    strategy="last",
    allow_partial=False,
    start_on="human",
    include_system=True,
    token_counter=local_token_counter  # Point to your local function!
)

messages_ds = [
    SystemMessage(content="You are an expert career coach specializing in AI and Data Science. Keep answers concise and actionable."),
    HumanMessage(content="Hi, I am Yash. I am a Data Scientist looking to transition into an LLM Engineer role."),
    AIMessage(content="Hello Yash! That is a hot transition right now. To start, you should master vector databases and fine-tuning frameworks like LoRA. What specific area do you want to focus on first?"),
    HumanMessage(content="I want to focus on building production-ready RAG pipelines. What tools should I start learning?")
]
trimmer.invoke(messages_ds)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough
chain=(
    RunnablePassthrough.assign(message=itemgetter("message")|trimmer)
    |prompt
    |model
)
```

```python
chain.invoke({"message":messages_ds,"language":"English"})
```
