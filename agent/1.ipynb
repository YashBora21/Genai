```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
import os
from  dotenv import load_dotenv
load_dotenv()
```

```python
llm = init_chat_model(
    model="gemma4:31b-cloud",
    model_provider="ollama",
    
)
```

```python
os.environ['Groq_API_KEY']=os.getenv('Groq_API_KEY')
```

```python
def getwhether(city:str)->str:
    """this will return the current whether"""
    return f"the whether in {city} is Sunny"

agent=create_agent(
    model=llm,
    tools=[getwhether],
    system_prompt="You are an helpful assistant"


)
agent
```

```python
respose=agent.invoke({"messages":[{"role":"user","content":"what is  the whether in city New York"}]})
```

```python
print(respose['messages'][-1])
```
