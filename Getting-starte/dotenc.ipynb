```python
import  os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['GEMINI_API_KEY']=os.getenv("GEMINI_API_KEY")
os.environ['LANGCHAIN_API_KEY']=os.getenv("langchain_api_key")
os.environ['LANGCHAIN_TRACING_V2']="true"
os.environ['LANGCHAIN_PROJECT']=os.getenv("langchain_project")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.0-flash")
print(llm)
```

```python
query="What is the capital of India?"
llm.invoke(query)
```

```python
from langchain_ollama import ChatOllama
llm2=ChatOllama(model="gemma4:31b-cloud")
print(llm2)
```

```python
query="What is the capital of India?"
llm2.invoke(query)
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt=ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant that translates English to hinglish."),
        ("human", "Translate the following English text to hinglish: {text}"),
    ]
)
```

```python
prompt
```

```python
chain=prompt|llm2
chain.invoke({"text":"hi my name is yash i an here to roar"})
```

```python
##stroutput  parser
from langchain_core.output_parsers import StrOutputParser
str_parser=StrOutputParser()
chain=prompt|llm2|str_parser
output=chain.invoke({"text":"hi my name is yash i an here to roar"})
```
